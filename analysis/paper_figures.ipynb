# Paper 2 figures — one panel per cell

State: full-horizon campaign, 10 seeds per configuration and geometry (120 runs), layout `runs_fullhorizon/{S1,T2,T4}/<geom>_<config>_s<seed>/<ckpt>`.
Style: no figure titles or subtitles except topology labels; no in-figure text except Fig. 1(a) and Fig. 2(a) (switchable per panel through `annotate`).

Run from `analysis/` (the repository root is `..`; override with the environment variable `P2_BUNDLE` if the checkpoints live elsewhere).
Every paper figure is split into **panel cells**: each cell holds the tweakable parameters of one panel in a dictionary
`P[...]`, draws that panel alone for a quick look, and the **assemble** cell at the end of each section stitches the panels
into the paper figure with exactly those parameters and writes the PDF into `../figs/`.

Heavy data (checkpoint evaluations) is computed once and cached in memory, so re-running a panel cell is instant.
Colours and linestyles per configuration live in `STYLE` (setup cell); font family and sizes in `FONT` (first code cell).

```python
# ======================= GLOBAL FONT KNOB (edit, re-run this cell, then re-run any panel/assemble cell) =======================
# Every figure in this notebook derives its font sizes from FONT.  'base' is the reference size (pt) at scale 1;
# the other entries are multipliers applied to it: axis labels, tick labels, legends, panel titles, and the in-figure
# annotations (whose individual sizes in the code are given relative to a 14 pt base and rescaled through FS()).
# Some sections use a larger overall scale because the figure is printed smaller (losses: 1.36; Fig. 1: 1.07).
import matplotlib.pyplot as plt

# ======================= GLOBAL STYLE OF THE ANALYTIC V* (edit, re-run this cell, then re-run any panel/assemble cell) ==========
# Every curve of the analytic solution (profiles, angular averages, S^1 maps, fold evolution) and its legend entry use VSTAR.
VSTAR_USER = dict(color='gray', lw=7, ls='-', alpha=0.7, zorder=1)   # any Line2D keyword: color, lw, ls ('-', '--', '-.', ':'), alpha, zorder, marker, ...
VSTAR_PRO = dict(color='k', lw=4.5, ls='-', alpha=0.9, zorder=1)      # journal style: solid black reference under the coloured curves

# ======================= PROFESSIONAL (JOURNAL) STYLE ===========================================================================
# PRO['on'] = True applies a journal look to every figure: Times-like text and math (STIX), thin inward ticks with minor ticks,
# thin spines, a recessive solid light-grey grid, frameless legends, and (palette=True) a colour-blind-safe palette with one
# linestyle per configuration.  Line and tick widths are scaled with the figure so that they print at the stated size.
# Everything is reversible: PRO['on'] = False gives back the previous look, STYLE_USER / VSTAR_USER keep your own choices.
PRO = dict(on=True,
           family='STIXGeneral', mathtext='stix',   # text and math faces (STIX = Times-like); e.g. 'DejaVu Serif' + 'dejavuserif', or 'sans-serif' + 'dejavusans'
           palette=True,                            # STYLE = STYLE_PRO (validated palette) instead of STYLE_USER
           vstar=True,                              # VSTAR = VSTAR_PRO instead of VSTAR_USER
           spine_lw=0.8, tick_dir='in', ticks_top_right=True, minor_ticks=True,   # axes chrome, in printed points
           grid_color='0.85', grid_lw=0.6,          # recessive solid grid
           legend_frame=False, handlelength=2.6)
VSTAR = VSTAR_USER if (PRO['on'] and PRO['vstar']) else VSTAR_USER



# -------------------- For figures 1, 2, -----------------------------
# FONT = dict(uniform=True,            # True: the base size is set per figure from the printed size: base = latex_pt * fig_width / print_width, so that a
#             #                        multiplier of 1 prints at latex_pt when the figure is scaled to print_width (\textwidth) in the paper ('base' is then unused);
#             #                        False: base = FONT['base'] * section scale.  In BOTH modes the multipliers below are applied on top of the base.
#             latex_pt=10.0,            # size the figure text must print at: the caption size of the paper (elsarticle: \small = 9 pt at the 10 pt option; use the body size, 10, if the captions are not reduced)
#             print_width=6.5,         # printed width of the figures in inches (\textwidth of the paper)
#             family='sans-serif',      # 'sans-serif' (DejaVu Sans), 'serif' (DejaVu Serif), 'Times New Roman', 'Helvetica', ...  (overridden by PRO['family'] while PRO['on'])
#             mathtext='dejavusans',    # 'dejavusans', 'dejavuserif', 'stix' (Times-like), 'cm' (Computer Modern)                 (overridden by PRO['mathtext'] while PRO['on'])
#             base=17.0,                # reference size in pt (uniform=False only)
#             label=1.6,                # axis labels          = base * label      (uniform=True: prints at latex_pt * label)
#             ticks=1.3,               # tick labels          = base * ticks
#             legend=1.3,               # legends              = base * legend
#             title=1.6,               # panel titles (topology labels, (a)/(b) letters) = base * title
#             annot=1.1)               # in-figure annotations: uniform=True: annot x base for every annotation; uniform=False: annot x the nominal size given in the code (relative to a 14 pt base) x FONT['base']/14 x scale



# -------------------- For figures 3, 4 5, and census-----------------------------

FONT = dict(uniform=True,            # True: the base size is set per figure from the printed size: base = latex_pt * fig_width / print_width, so that a
            #                        multiplier of 1 prints at latex_pt when the figure is scaled to print_width (\textwidth) in the paper ('base' is then unused);
            #                        False: base = FONT['base'] * section scale.  In BOTH modes the multipliers below are applied on top of the base.
            latex_pt=10.0,            # size the figure text must print at: the caption size of the paper (elsarticle: \small = 9 pt at the 10 pt option; use the body size, 10, if the captions are not reduced)
            print_width=6.5,         # printed width of the figures in inches (\textwidth of the paper)
            family='sans-serif',      # 'sans-serif' (DejaVu Sans), 'serif' (DejaVu Serif), 'Times New Roman', 'Helvetica', ...  (overridden by PRO['family'] while PRO['on'])
            mathtext='dejavusans',    # 'dejavusans', 'dejavuserif', 'stix' (Times-like), 'cm' (Computer Modern)                 (overridden by PRO['mathtext'] while PRO['on'])
            base=17.0,                # reference size in pt (uniform=False only)
            label=1.2,                # axis labels          = base * label      (uniform=True: prints at latex_pt * label)
            ticks=1.2,               # tick labels          = base * ticks
            legend=1.1,               # legends              = base * legend
            title=1.5,               # panel titles (topology labels, (a)/(b) letters) = base * title
            annot=0.95)   



_FS = dict(scale=1.0, uniform_pt=None)
def fonts(scale=None, fig_width=None):
    # apply FONT to matplotlib.  scale: section multiplier (kept from the last call when None); fig_width: width in inches of the
    # figure about to be drawn -- in uniform mode the base size becomes latex_pt * fig_width / print_width (and the chrome is scaled alike);
    # otherwise the base is FONT['base'] * scale.  The multipliers FONT['label'|'ticks'|'legend'|'title'] are applied in both modes.
    if scale is None: scale = _FS['scale']
    _FS['scale'] = scale
    fam, mt = (PRO['family'], PRO['mathtext']) if PRO['on'] else (FONT['family'], FONT['mathtext'])
    if FONT['uniform'] and fig_width is not None:
        S = fig_width / FONT['print_width']; b = FONT['latex_pt'] * S; _FS['uniform_pt'] = b
    else:
        S = scale; b = FONT['base'] * scale; _FS['uniform_pt'] = None
    pro_rc(S)
    plt.rcParams.update({'font.family': fam, 'mathtext.fontset': mt, 'font.size': b,
                         'axes.labelsize': b * FONT['label'], 'xtick.labelsize': b * FONT['ticks'], 'ytick.labelsize': b * FONT['ticks'],
                         'legend.fontsize': b * FONT['legend'], 'axes.titlesize': b * FONT['title'], 'figure.titlesize': b * FONT['title']})
def pro_rc(S):
    # axes chrome scaled by S (figure width / printed width), so that spines, ticks and grid print at PRO's point sizes
    if not PRO['on']:
        plt.rcParams.update({'axes.linewidth': 0.8, 'xtick.direction': 'out', 'ytick.direction': 'out', 'xtick.top': False, 'ytick.right': False,
                             'xtick.minor.visible': False, 'ytick.minor.visible': False, 'grid.color': '#b0b0b0', 'grid.linewidth': 0.8,
                             'legend.frameon': True, 'pdf.fonttype': 42}); return
    plt.rcParams.update({'axes.linewidth': PRO['spine_lw'] * S, 'xtick.direction': PRO['tick_dir'], 'ytick.direction': PRO['tick_dir'],
                         'xtick.top': PRO['ticks_top_right'], 'ytick.right': PRO['ticks_top_right'],
                         'xtick.minor.visible': PRO['minor_ticks'], 'ytick.minor.visible': PRO['minor_ticks'],
                         'xtick.major.size': 3.5 * S, 'ytick.major.size': 3.5 * S, 'xtick.major.width': PRO['spine_lw'] * S, 'ytick.major.width': PRO['spine_lw'] * S,
                         'xtick.minor.size': 2.0 * S, 'ytick.minor.size': 2.0 * S, 'xtick.minor.width': 0.6 * PRO['spine_lw'] * S, 'ytick.minor.width': 0.6 * PRO['spine_lw'] * S,
                         'xtick.major.pad': 3.5 * S, 'ytick.major.pad': 3.5 * S, 'axes.labelpad': 4.0 * S, 'axes.titlepad': 6.0 * S,
                         'grid.color': PRO['grid_color'], 'grid.linewidth': PRO['grid_lw'] * S, 'grid.linestyle': '-', 'grid.alpha': 1.0,
                         'legend.frameon': PRO['legend_frame'], 'legend.handlelength': PRO['handlelength'], 'legend.columnspacing': 1.6, 'legend.handletextpad': 0.6,
                         'lines.solid_capstyle': 'round', 'lines.dash_capstyle': 'round', 'pdf.fonttype': 42, 'savefig.dpi': 300})
def grid(ax, which='major'):
    # recessive grid: solid light-grey major lines, lighter and thinner minor lines
    ax.grid(True, which='major', color=plt.rcParams['grid.color'], lw=plt.rcParams['grid.linewidth'], ls='-', alpha=1.0 if PRO['on'] else 0.3)
    if which == 'both': ax.grid(True, which='minor', color='0.92' if PRO['on'] else plt.rcParams['grid.color'], lw=0.5 * plt.rcParams['grid.linewidth'], ls='-', alpha=1.0 if PRO['on'] else 0.3)
def FS(pt):
    # size of an in-figure annotation.  uniform mode: FONT['annot'] x the figure's base size (the nominal pt is ignored);
    # otherwise the nominal size given in the code (relative to a 14 pt base), rescaled by FONT['base'], the section scale and FONT['annot']
    if FONT['uniform'] and _FS['uniform_pt'] is not None: return _FS['uniform_pt'] * FONT['annot']
    return pt * (FONT['base'] / 14.0) * _FS['scale'] * FONT['annot']
fonts()
```

```python
import os, sys, json, math, itertools
import numpy as np, torch
import matplotlib; import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
from matplotlib.colors import ListedColormap

B = os.environ.get('P2_BUNDLE', os.path.abspath('..'))     # repository root (gpu_critic.py, runs_fullhorizon/, fold_evolution/)
FIGS = os.path.abspath('../figs'); os.makedirs(FIGS, exist_ok=True)
sys.path.insert(0, B)
import gpu_critic as gc, grid_certificate as se
torch.set_num_threads(4)
EPS, DV = 0.08, 0.2

# ---- configurations: internal key -> paper name, colour, linestyle ------------------------------
ARMS = ['smooth', 'rawsoft', 'min', 'folded']
# STYLE = {'smooth':  dict(label='Fourier-feature', color='#c0392b', ls='-'),
#          'rawsoft': dict(label='soft-periodic',   color='#2ca02c', ls='--'),
#          'min':     dict(label='min-of-charts',   color='#1f77b4', ls='-.'),
#          'folded':  dict(label='geometric',       color='#ff7f0e', ls=':')}

# 'lw' and 'zorder' given here override the per-section defaults (P[...]['lw']) for every curve and legend entry of that configuration
STYLE_USER = {'smooth':  dict(label='Fourier-feature', color='#c0392b', ls='-', lw=2, zorder=5),
         'rawsoft': dict(label='soft-periodic',   color='green', ls='-.', lw=2.5, zorder=2),
         'min':     dict(label='min-of-charts',   color='orange', ls='-', lw=2.75, zorder=3),
         'folded':  dict(label='geometric',       color='blue', ls='--', lw=1.5, zorder=4)}

# Journal palette: validated colour-blind-safe (adjacent and all-pairs CVD dE >= 8.4, normal-vision dE >= 17.5; the amber sits at 3:1 on white),
# one linestyle per configuration as the secondary encoding; the correct solutions are drawn on top of the Fourier-feature one.
STYLE_PRO = {'smooth':  dict(label='Fourier-feature', color='#c0392b', ls='-',  lw=2.0, zorder=2),
             'rawsoft': dict(label='soft-periodic',   color='#199e70', ls='--', lw=2.2, zorder=3),
             'min':     dict(label='min-of-charts',   color='#2a78d6', ls='-.', lw=2.2, zorder=4),
             'folded':  dict(label='geometric',       color='#c98500', ls=':',  lw=2.6, zorder=5)}


STYLE = STYLE_USER if (PRO['on'] and PRO['palette']) else STYLE_USER
GEOM = {'torus': gc.Torus, 'circle': gc.Circle, 'torus4': gc.Torus4}
GTITLE = {'torus': 'torus $\\mathbb{T}^{\\ \\,2}$', 'circle': 'circle $S^1$', 'torus4': 'torus $\\mathbb{T}^{\\ \\,4}$'}
ITERS = {'torus': '$10^6$ iterations', 'circle': '$5\\times10^5$ iterations', 'torus4': '$10^6$ iterations'}

# ---- checkpoints: runs_fullhorizon/{S1,T2,T4}/<geom>_<config>_s<seed>/<ckpt>, every seed found on disk (up to P2_SEEDS, default 10)
NSEEDS = int(os.environ.get('P2_SEEDS', 10))
SUB = {'circle': 'S1', 'torus': 'T2', 'torus4': 'T4'}
CK = {'smooth': 'gpu_{g}.pt', 'rawsoft': 'gpu_{g}_rawsoft.pt', 'min': 'gpu_{g}_charts.pt', 'folded': 'gpu_{g}_charts1.pt'}
def _ckpt(geom, arm, seed): return f"{B}/runs_fullhorizon/{SUB[geom]}/{geom}_{arm}_s{seed}/" + CK[arm].format(g=geom)
RUNS = {g: {a: [p for p in (_ckpt(g, a, sd) for sd in range(NSEEDS)) if os.path.exists(p)] for a in ARMS} for g in GEOM}
SEEDS = {g: {a: [int(p.split('_s')[-1].split('/')[0]) for p in RUNS[g][a]] for a in ARMS} for g in GEOM}   # seed numbers, in order
# the flat consolidated layout runs_fullhorizon/<geom>_<config>_s<seed> of the 4-seed stage is accepted as a fallback
for g in GEOM:
    for a in ARMS:
        if not RUNS[g][a]:
            RUNS[g][a] = [p for p in (f"{B}/runs_fullhorizon/{g}_{a}_s{sd}/" + CK[a].format(g=g) for sd in range(NSEEDS)) if os.path.exists(p)]
            SEEDS[g][a] = [int(p.split('_s')[-1].split('/')[0]) for p in RUNS[g][a]]
def chunked(fwd, d, t, n=20000):
    with torch.no_grad(): return torch.cat([fwd(d[i:i + n], t[i:i + n]) for i in range(0, d.shape[0], n)])
def vstar(dist, tau): return np.minimum(tau, np.maximum(dist - EPS, 0) / DV)
def eps_xticks(ax, xmax):
    """x axis from the capture radius: xlim = (EPS, xmax), with a tick labelled epsilon at the left end."""
    ax.set_xlim(EPS, xmax); t = [v for v in ax.get_xticks() if EPS + 0.06 < v <= xmax + 1e-9]
    ax.set_xticks([EPS] + t); ax.set_xticklabels(['$\\varepsilon$'] + [f'{v:g}' for v in t])

_models = {}
def load(geom, arm, seed=0):
    """-> (G, model, fwd(delta, tau), charts_flag, checkpoint dict); cached."""
    key = (geom, arm, seed)
    if key in _models: return _models[key]
    G = GEOM[geom](); D = G.features(G.sample_pos(2, 'cpu'), torch.zeros(2)).shape[1]
    if arm == 'smooth':   m = gc.MLP(D, 128); fwd = lambda d, t: m(G.features(d, t)); ch = False
    elif arm == 'rawsoft': m = gc.RawSoftCritic(G, 128); fwd = lambda d, t: m.forward_torus(d, t); ch = True
    else:                 m = gc.ChartsCritic(G, D, 128, single=(arm == 'folded')); fwd = lambda d, t: m.forward_torus(d, t); ch = True
    ck = torch.load(RUNS[geom][arm][seed], map_location='cpu'); m.load_state_dict(ck['model']); m.eval()
    _models[key] = (G, m, fwd, ch, ck); return _models[key]

def hist(geom, arm, seed):
    return torch.load(RUNS[geom][arm][seed], map_location='cpu')['hist']

def medfilt(v, w=9):
    v = np.asarray(v, float); out = np.copy(v)
    for i in range(len(v)): out[i] = np.median(v[max(0, i - w // 2):i + w // 2 + 1])
    return out

P = {}            # all panel parameters live here
def legend_handles(counts=None, vstar_line=True, lw=2, ls=None):
    h = [Line2D([], [], label='analytic $V^*$', **VSTAR)] if vstar_line else []
    for k in ARMS:
        lab = STYLE[k]['label'] #+ (f"  ({counts[k]} seed{'s' if counts[k] > 1 else ''})" if counts else '')
        h.append(Line2D([], [], color=STYLE[k]['color'], ls=(ls or STYLE[k]['ls']), lw=STYLE[k].get('lw', lw), label=lab))
    return h
print('bundle:', B); print('figures ->', FIGS); print('seeds per configuration:', {g: {a: len(RUNS[g][a]) for a in RUNS[g]} for g in RUNS})
missing = {g: {a: sorted(set(range(NSEEDS)) - set(SEEDS[g][a])) for a in ARMS if len(SEEDS[g][a]) < NSEEDS} for g in GEOM}
print('missing seeds:', {g: m for g, m in missing.items() if m})
```

## Figure 1 — analytic value function $V^*$

```python
TMAX, DMAX = 8.0, np.sqrt(2)
P['f1a'] = dict(cmap='viridis', cone_color='#157a4f', crease_color='#b3261e', title='', annotate=True, exit_text_x=0.095)   # title: panel letter, e.g. '(a)', if wanted; exit_text_x: x of the 'cone reaches the far corner' text (was 0.12)
def panel_f1a(a, p=P['f1a']):
    d = np.linspace(0, DMAX, 600); t = np.linspace(0, TMAX, 600); D, T = np.meshgrid(d, t)
    V = np.minimum(T, np.maximum(D - EPS, 0) / DV); V[D <= EPS] = np.nan
    im = a.pcolormesh(D, T, V, shading='auto', cmap=p['cmap'], rasterized=True)
    a.axvspan(0, EPS, color='k'); a.text(EPS / 2, 4, 'capture set  $V\\equiv 0$', rotation=90, color='w', ha='center', va='center', fontsize=FS(12), fontweight='bold')
    tc = np.linspace(0, (DMAX - EPS) / DV, 2); a.plot(EPS + DV * tc, tc, color='w', lw=4); a.plot(EPS + DV * tc, tc, color=p['cone_color'], lw=2)
    tau_exit = (DMAX - EPS) / DV; a.axhline(tau_exit, color='w', ls=':', lw=1.5)
    a.axvline(1.0, color='w', ls='--', lw=1.5); a.fill_betweenx([0, TMAX], 1.0, DMAX, facecolor='none', hatch='///', edgecolor='w', alpha=0.35, lw=0)
    if p['annotate']:
        a.text(p['exit_text_x'], tau_exit + 0.15, f'$\\tau={tau_exit:.2f}$', color='w', fontsize=FS(11))
        # a.text(0.45, 4.55, 'distance-limited\n$V=(d-\\varepsilon)/\\Delta v$\n$\\partial_\\tau V=0,\\ \\|\\nabla V\\|=1/\\Delta v$', color='w', ha='center', fontsize=FS(12.5))
        # a.text(0.78, 1.2, 'horizon-limited\n$V=\\tau$\n$\\partial_\\tau V=1,\\ \\nabla V=0$', color='w', ha='center', fontsize=FS(12.5))
        a.text(0.45, 4.55, 'distance-limited\n$V=(d-\\varepsilon)/\\Delta v$', color='w', ha='center', fontsize=FS(12.5))
        a.text(0.78, 1.2, 'horizon-limited\n$V=\\tau$', color='w', ha='center', fontsize=FS(12.5))

        a.annotate('cone  $d=\\varepsilon+\\Delta v\\,\\tau$\n(concave kink)', xy=(EPS + DV * 3.4, 3.4), xytext=(0.4, 2.45), color=p['cone_color'], fontsize=FS(12.5), fontweight='bold', ha='center',
                   bbox=dict(boxstyle='round', fc='w', ec=p['cone_color'], alpha=0.85), arrowprops=dict(arrowstyle='->', color='darkgreen', lw=1.5))
        a.annotate('capture crease\n(convex kink)', xy=(EPS, 0.9), xytext=(0.36, 0.4), color=p['crease_color'], fontsize=FS(12), fontweight='bold', ha='center',
                   bbox=dict(boxstyle='round', fc='w', ec=p['crease_color'], alpha=0.85), arrowprops=dict(arrowstyle='->', color=p['crease_color'],lw=1.5))
        a.text(1.20, 3.45, 'ridge\n(cut locus)\n$d\\in[1,\\sqrt{2}]$', color='k', fontsize=FS(11.5), ha='center', va='center', fontweight='bold', bbox=dict(boxstyle='round', fc='w', ec='0.3', alpha=0.85))
    a.set_xlabel('min-image distance  $d$'); a.set_ylabel('remaining horizon  $\\tau$'); a.set_xlim(0, DMAX); a.set_ylim(0, TMAX)
    a.set_title(p['title'], loc='left')
    return im
fonts(scale=1.07)
fig, a = plt.subplots(figsize=(8.5, 5.8)); im = panel_f1a(a); fig.colorbar(im, ax=a, pad=0.01).set_label('$V^*$'); plt.show()
```

## Figure 2 — chart selector on $\mathbb{T}^2$

Drawn on the fundamental domain $[0,2)^2$ with the pursuer at the corners (the displacement shifted by half a period).

```python
TAU_SEL = 4.0
IMGS = np.array([[0, 0], [2, 0], [0, 2], [2, 2]]); IMG_NAMES = ['image (0,0): straight across', 'image (2,0): wrap in $x$', 'image (0,2): wrap in $y$', 'image (2,2): wrap in both']
SEL_COLS = ['#cfe2f3', '#fbe0c4', '#d9ead3', '#f4cccc']
def selector_data(n=401, seed=0):
    xs = np.linspace(0, 2, n); X, Y = np.meshgrid(xs, xs); Pp = np.stack([X, Y], -1)
    dist = np.linalg.norm(Pp[..., None, :] - IMGS[None, None], axis=-1); m_an = dist.argmin(-1); d = dist.min(-1)
    G, model, fwd, _, _ = load('torus', 'min', seed); W = ((Pp + 1) % 2) - 1
    with torch.no_grad():
        pt = torch.tensor(W.reshape(-1, 2), dtype=torch.float32); tt = torch.full((pt.shape[0],), TAU_SEL)
        outs = [model.net(model.ext_features(pt + sh, tt)) for sh in model._image_shifts(pt)]
        st = torch.stack(outs, 0); am = st.argmin(0).numpy().reshape(n, n)
        shifts = [sh.numpy().reshape(n, n, 2) for sh in model._image_shifts(pt)]
    sel = np.zeros((n, n), int)
    for k in range(4):
        img = np.rint((Pp - (W + shifts[k])) / 2) * 2
        idx = (img[..., 0] > 1).astype(int) + 2 * (img[..., 1] > 1).astype(int); sel[am == k] = idx[am == k]
    hor = vstar(d, TAU_SEL) >= TAU_SEL - 1e-6
    agree = (sel == m_an)[~hor].mean() * 100
    return dict(X=X, Y=Y, m_an=m_an, d=d, sel=sel, hor=hor, agree=agree)
SD = selector_data(); print('selector agreement on distance-limited states: %.2f%%' % SD['agree'])
```

```python
P['f2a'] = dict(title='', quiver_n=12, arrow_color='#2b3440', annotate=True)
def panel_f2a(ax, p=P['f2a']):
    X, Y = SD['X'], SD['Y']; cm = ListedColormap(SEL_COLS)
    ax.pcolormesh(X, Y, SD['m_an'], cmap=cm, vmin=-0.5, vmax=3.5, shading='auto', rasterized=True)
    ax.axvline(1, color='k', lw=2.5); ax.axhline(1, color='k', lw=2.5)
    q = np.linspace(0.08, 1.92, p['quiver_n']); QX, QY = np.meshgrid(q, q); QP = np.stack([QX, QY], -1)
    qd = np.linalg.norm(QP[..., None, :] - IMGS[None, None], axis=-1); qm = qd.argmin(-1); head = QP - IMGS[qm]; head /= np.linalg.norm(head, axis=-1, keepdims=True)
    ax.quiver(QX, QY, head[..., 0], head[..., 1], color=p['arrow_color'], scale=22, width=0.006)
    if p['annotate']:
        for k, (cx, cy) in enumerate([(0.5, 0.5), (1.5, 0.5), (0.5, 1.5), (1.5, 1.5)]):
            ax.text(cx, cy, IMG_NAMES[k].replace(': ', ':\n'), ha='center', va='center', fontsize=FS(15), fontweight='bold', bbox=dict(fc='w', ec='none', alpha=0.8))
        ax.text(0.04, 0.04, 'pursuer', fontsize=FS(11), va='bottom', bbox=dict(fc='w', ec='none', alpha=0.8))
        ax.text(1.0, 1.95, 'cut locus of $d$ (ridge)', ha='center', va='top', fontsize=FS(15), fontweight='bold', bbox=dict(fc='w', ec='none', alpha=0.85))
    for c in IMGS: ax.plot(*c, 'o', ms=14, color='k', clip_on=False, zorder=5)
    ax.set_xlim(0, 2); ax.set_ylim(0, 2); ax.set_aspect('equal'); ax.set_xlabel('$\\Delta_x$'); ax.set_ylabel('$\\Delta_y$'); ax.set_title(p['title'], loc='left')
fonts(scale=1)
fig, ax = plt.subplots(figsize=(6.5, 6.2)); panel_f2a(ax); plt.show()
```

## Figures 4 and 5 — profiles at $\tau=3$ (value and $|V-V^*|$)

Profiles are evaluated once per (path, configuration, seed) and cached.  Five paths: `circle`, `torus_diag`, `torus_axis`,
`torus4_diag`, `torus4_axis`.  One cell per panel (path × {value, error}); the assemble cells build the paper layouts:
the 3-row and 5-row figures (one geometry per row, value | error), the earlier two-column layouts, and the single-axis T² figure.
`P['prof']['mark_failed']` draws the relabelled min-of-charts seeds (`FAILED`) in dark blue; `P['prof']['annotate']` / `['letters']` restore the cone text and lettered subtitles.

```python
TAU_PROF = 3.0; OFF = 0.02
PATHS = {'circle':      dict(geom='circle', P_of=lambda s: s[:, None], xlab='$d(\\Delta)$', title='circle $S^1$'),
         'torus_diag':  dict(geom='torus',  P_of=lambda s: np.stack([s, s], 1), xlab='$d$ along the diagonal $\\Delta=(s,s)$', title='torus $\\mathbb{T}^{\\ \\,2}$, diagonal'),
         'torus_axis':  dict(geom='torus',  P_of=lambda s: np.stack([s, 0 * s + OFF], 1), xlab='$d$ along one axis $\\Delta=(s,0.02)$', title='torus $\\mathbb{T}^{\\ \\,2}$, single axis'),
         'torus4_diag': dict(geom='torus4', P_of=lambda s: np.stack([s, s, s, s], 1), xlab='$d$ along the main diagonal $\\Delta=(s,s,s,s)$', title='torus $\\mathbb{T}^{\\ \\,4}$, main diagonal'),
         'torus4_axis': dict(geom='torus4', P_of=lambda s: np.stack([s, 0 * s + OFF, 0 * s + OFF, 0 * s + OFF], 1), xlab='$d$ along one axis $\\Delta=(s,0.02,0.02,0.02)$', title='torus $\\mathbb{T}^{\\ \\,4}$, single axis')}
FAILED = {('torus', 'min', 2), ('torus', 'min', 8), ('torus4', 'min', 1)}      # relabelled image (selector never picks the identity), degraded value; indices into RUNS
_prof = {}
def profile(path_key, arm, seed, n=1200, tau=TAU_PROF):
    key = (path_key, arm, seed)
    if key in _prof: return _prof[key]
    spec = PATHS[path_key]; geom = spec['geom']
    s = np.linspace(0.0, 0.999, n); Pm = spec['P_of'](s); dist = np.linalg.norm(Pm, axis=1); Vs = vstar(dist, tau)
    G, m, fwd, ch, _ = load(geom, arm, seed)
    v = chunked(fwd, torch.tensor(Pm, dtype=torch.float32), torch.full((n,), tau)).numpy(); err = np.abs(v - Vs)
    v[dist <= EPS] = np.nan; err[dist <= EPS] = np.nan
    _prof[key] = dict(dist=dist, v=v, err=err, Vs=Vs); return _prof[key]

P['prof'] = dict(lw=1.6, lw_err=1.6, alpha=0.9, vstar_lw=4, ylim_value=(-0.3, 3.6), ylim_err=(1e-5, 10), cone_color='0.5', fontsize_title=14.5,
                 mark_failed=False, dark='#08306b', annotate=False, letters=False, grid_left=False, grid_right=False)
def _title_pt(p):
    # size of the profile titles: the figure's title size (FONT['title'] x base) in uniform mode, else the nominal fontsize_title through FS()
    return plt.rcParams['axes.titlesize'] if _FS['uniform_pt'] is not None else FS(p['fontsize_title'])
def panel_profile(ax, path_key, row, letter='', p=P['prof'], seeds=None):
    spec = PATHS[path_key]; geom = spec['geom']; dpl = EPS + DV * TAU_PROF
    ref = profile(path_key, 'folded', 0); dist = ref['dist']
    if row == 'value': ax.plot(dist, ref['Vs'], **VSTAR)
    for arm in ARMS:
        for s in (seeds if seeds is not None else range(len(RUNS[geom][arm]))):
            pr = profile(path_key, arm, s); bad = p['mark_failed'] and (geom, arm, s) in FAILED
            kw = dict(color=p['dark'] if bad else STYLE[arm]['color'], ls=STYLE[arm]['ls'], lw=2.2 if bad else STYLE[arm].get('lw', p['lw']), alpha=0.95 if bad else p['alpha'], zorder=3 if bad else STYLE[arm].get('zorder', 2))
            if row == 'value': ax.plot(dist, pr['v'], **kw)
            else: ax.semilogy(dist, pr['err'] + 1e-9, **{**kw, 'lw': p['lw_err']})
    ax.axvline(dpl, color=p['cone_color'], ls=':', lw=1.5); ax.set_xlabel(spec['xlab']); eps_xticks(ax, dist.max())
    if row == 'value':
        if p['grid_left']: grid(ax, 'major')
        if p['annotate']: ax.text(dpl + 0.02 * dist.max(), 0.9, 'cone: $d=\\varepsilon+\\Delta v\\,\\tau$\nplateau $V^*=\\tau$ beyond', fontsize=FS(11), color='0.3')
        ax.set_ylabel(f'$V(d(\\Delta),\\tau={TAU_PROF:g})$'); ax.set_ylim(*p['ylim_value'])
        ax.set_title(f"({letter})  {spec['title']}: value at $\\tau={TAU_PROF:g}$" if p['letters'] else spec['title'], loc='left', fontsize=_title_pt(p))
    else:
        if p['grid_right']: grid(ax, 'both')
        ax.set_ylim(*p['ylim_err'])
        ax.set_ylabel('$|V-V^*|$')
        if p['letters']: ax.set_title(f"({letter})  {spec['title']}: absolute error", loc='left', fontsize=_title_pt(p))
def profile_legend(fig, geoms, ncol=2, seeds=None):
    if seeds is not None and len(seeds) == 1: H = legend_handles(); H = H[:1] + [Line2D([], [], color=STYLE[k]['color'], ls=STYLE[k]['ls'], lw=STYLE[k].get('lw', 2), label=f"{STYLE[k]['label']} NN (seed {seeds[0]})") for k in ARMS]
    else: H = legend_handles({a: (len(seeds) if seeds is not None else max(len(RUNS[g][a]) for g in geoms)) for a in ARMS})
    if P['prof']['mark_failed'] and any((g, 'min', s) in FAILED for g in geoms for s in range(10)):
        H.insert(4, Line2D([], [], color=P['prof']['dark'], ls=STYLE['min']['ls'], lw=2.4, label='min-of-charts, minimum inactive'))
    fig.legend(handles=H, loc='upper center', ncol=ncol, bbox_to_anchor=(0.5, 1.0), frameon=False, fontsize=FS(12.5))
fonts(scale=1)
```

## Figures 6 and 7 — structure maps at $\tau=6$ (seed 0)

One cell per configuration row (value, gradient norm, log residual, error). The $\mathbb{T}^4$ figure is the slice $\Delta_3=\Delta_4=0.02$.

```python
TAU_MAP, NMAP, OFF = 4.7, 301, 0.02
_maps = {}
def mapdata(geom, arm, seed=0):
    key = (geom, arm, seed)
    if key in _maps: return _maps[key]
    xs = np.linspace(-0.99, 0.99, NMAP); X, Y = np.meshgrid(xs, xs)
    Pm = np.stack([X.ravel(), Y.ravel()], 1) if geom == 'torus' else np.stack([X.ravel(), Y.ravel(), np.full(X.size, OFF), np.full(X.size, OFF)], 1)
    d = np.linalg.norm(Pm, axis=1).reshape(NMAP, NMAP); Vs = vstar(d, TAU_MAP)
    G, m, fwd, ch, _ = load(geom, arm, seed)
    pt = torch.tensor(Pm, dtype=torch.float32); tt = torch.full((pt.shape[0],), TAU_MAP)
    Vl, gl, rl = [], [], []
    for k in range(0, pt.shape[0], 10000):
        V, gd, gt = gc.eval_V_and_grads(m, G, ch, pt[k:k + 10000].clone(), tt[k:k + 10000].clone())
        rl.append(G.residual_terms(V, gd, gt, 1.0, 1.2).detach()); Vl.append(V.detach()); gl.append(gd.detach().norm(dim=-1))
    V = torch.cat(Vl).numpy().reshape(NMAP, NMAP); gn = torch.cat(gl).numpy().reshape(NMAP, NMAP); r = torch.cat(rl).numpy().reshape(NMAP, NMAP)
    cap = d <= EPS
    _maps[key] = dict(X=X, Y=Y, V=np.where(cap, np.nan, V), gn=np.where(cap, np.nan, gn), logr=np.where(cap, np.nan, np.log10(np.abs(r) + 1e-12)), err=np.where(cap, np.nan, np.abs(V - Vs)))
    return _maps[key]

P['maps'] = dict(cmap='viridis', vmax_V=5, vmax_grad=6, logr_range=(-6, 0.5), cone_color='r', cone_ls='--',
                 errmax={('torus', 'smooth'): 4, ('torus', 'rawsoft'): 0.1, ('torus', 'min'): 0.03, ('torus', 'folded'): 0.03,
                         ('torus4', 'smooth'): 4, ('torus4', 'rawsoft'): 1.0, ('torus4', 'min'): 0.05, ('torus4', 'folded'): 0.05},
                 cols=['$V_{\\mathrm{NN}}$', '$\\|\\nabla_\\Delta V\\|$', '$\\log_{10}|r|$', '$|V_{\\mathrm{NN}}-V^*|$'], suptitle=False,
                 cbar_label_scale=1.4,    # KNOB: size of the colorbar labels (the former column titles) relative to the other text of the figure
                                          # (tick numbers, axis labels); 1.0 = same size.  Tick labels of the colorbars follow the axis ticks.
                 cbar_label_size=20,      # used only when FONT['uniform'] is False (nominal size at a 14 pt base, through FS())
                 cbar_labelpad=14,        # KNOB: gap between the colorbar tick labels and the colorbar label, in printed points
                 axis_labels={'torus': ('$\\Delta_x$', '$\\Delta_y$'), 'torus4': ('$\\Delta_1$', '$\\Delta_2$')},   # x / y axis labels (bottom row, left column)
                 ylabel_pad=2.0,          # KNOB: gap between the left-most panel and its y label ($\Delta_y$), in printed points (figure default: 4)
                 row_label_x=-0.40,       # position of the bold configuration name, in axes fraction left of the left-most panel (more negative = further left)
                 row_label_scale=1.3,     # KNOB: size of the bold configuration names relative to the base text size of the figure (uniform mode); 1.0 = tick-label size
                 cbar_height=0.06,        # colorbar thickness, as a fraction of the panel height
                 cbar_pad=0.04,           # gap between panel and colorbar, as a fraction of the panel height
                 err_shared=None)         # None: the error column keeps its per-row limits (errmax) and gets one colorbar per row;
                                          # a number: common vmax for the whole error column and a single colorbar, like the other columns
def _ps():
    # drawn points per printed point of the current figure (uniform mode), so that pads given in printed points print as stated
    return _FS['uniform_pt'] / FONT['latex_pt'] if (FONT['uniform'] and _FS['uniform_pt'] is not None) else 1.0
def _row_label_pt(p):
    # size of the row labels: row_label_scale x base in uniform mode, else the former nominal 16 pt through FS()
    return _FS['uniform_pt'] * p['row_label_scale'] if _FS['uniform_pt'] is not None else FS(16)
def _hcbar(fig, a, im, p, label=None, ticks=None, ticklabels=None):
    cax = a.inset_axes([0, 1 + p['cbar_pad'], 1, p['cbar_height']])
    cb = fig.colorbar(im, cax=cax, orientation='horizontal')
    cb.ax.xaxis.set_ticks_position('top'); cb.ax.xaxis.set_label_position('top')
    cb.ax.tick_params(labelsize=plt.rcParams['xtick.labelsize'])
    if ticks is not None: cb.set_ticks(ticks); cb.set_ticklabels(ticklabels)
    if label:
        base = _FS['uniform_pt'] if (FONT['uniform'] and _FS['uniform_pt'] is not None) else FS(p['cbar_label_size'])
        S = base / FONT['latex_pt'] if (FONT['uniform'] and _FS['uniform_pt'] is not None) else 1.0
        cb.set_label(label, fontsize=base * p['cbar_label_scale'], labelpad=p['cbar_labelpad'] * S)
    return cb
def panel_maprow(axrow, fig, geom, arm, p=P['maps'], titles=True, first=True, last=True):
    """One configuration row.  first: this is the top row (carries the shared colorbars of columns 0-2 and the labels);
    last: bottom row (x tick labels).  y tick labels only on the left-most panel."""
    M = mapdata(geom, arm); emax = p['err_shared'] if p['err_shared'] is not None else p['errmax'][(geom, arm)]
    panels = [M['V'], M['gn'], M['logr'], M['err']]
    kws = [dict(vmin=0, vmax=p['vmax_V']), dict(vmin=0, vmax=p['vmax_grad']), dict(vmin=p['logr_range'][0], vmax=p['logr_range'][1]), dict(vmin=0, vmax=emax)]
    R = EPS + DV * TAU_MAP; Rs = np.sqrt(max(R ** 2 - (2 * OFF ** 2 if geom == 'torus4' else 0), 0)); th = np.linspace(0, 2 * np.pi, 400)
    for j in range(4):
        a = axrow[j]; im = a.pcolormesh(M['X'], M['Y'], panels[j], shading='auto', rasterized=True, cmap=p['cmap'], **kws[j])
        a.plot(Rs * np.cos(th), Rs * np.sin(th), color=p['cone_color'], ls=p['cone_ls'], lw=1.3)
        a.set_xlim(-1, 1); a.set_ylim(-1, 1); a.set_aspect('equal'); a.set_xticks([-1, 0, 1]); a.set_yticks([-1, 0, 1])
        a.tick_params(labelbottom=last, labelleft=(j == 0))
        if last: a.set_xlabel(p['axis_labels'][geom][0])
        if j == 0: a.set_ylabel(p['axis_labels'][geom][1], labelpad=p['ylabel_pad'] * _ps())
        shared = (j < 3) or (p['err_shared'] is not None)
        if shared and first:
            _hcbar(fig, a, im, p, label=p['cols'][j] if titles else None)
        elif not shared:
            tk = np.linspace(0, emax, 3) if emax <= 1.0 else None
            _hcbar(fig, a, im, p, label=(p['cols'][j] if (titles and first) else None), ticks=tk, ticklabels=None if tk is None else [f'{t:.3g}' for t in tk])
    axrow[0].text(p['row_label_x'], 0.5, STYLE[arm]['label'], transform=axrow[0].transAxes, rotation=90, ha='center', va='center', fontsize=_row_label_pt(p), fontweight='bold')
fonts(scale=1.07)
```

## $V(d,\tau)$ pictures — analytic $V^*$ and the trained NNs (one seed)

On $S^1$ the $(d,\tau)$ plane is the function itself.  On the tori the NN also depends on the direction of $\Delta$, so each
configuration gets a cut along one axis, a cut along the main diagonal, and the spread $\max_u V-\min_u V$ over directions

## Angular mean and spread on $\mathbb{T}^2$ (one seed, $\tau=3$)

At each distance $d$, $V$ is evaluated on the circle $|\Delta|=d$ (directions inside the fundamental domain) and reduced to its
angular mean and standard deviation; the mean is compared with $V^*$ and the standard deviation measures the departure from a
function of $d$ alone.

```python
P['angavg'] = dict(tau=3.0, seed=0, nth=720, lim=0.999, band_alpha=0.18, lw=2, lw_err=2, ylim_value=(-0.3, 3.6), ylim_std=(1e-5, 10), annotate=False, grid_left=False)
def angavg_fig(fname=None, p=P['angavg']):
    tau, seed = p['tau'], p['seed']
    d = np.linspace(0.0, np.sqrt(2) * p['lim'], 700); th = (np.arange(p['nth']) + 0.5) * 2 * np.pi / p['nth']
    X = d[:, None] * np.cos(th)[None, :]; Y = d[:, None] * np.sin(th)[None, :]; inside = (np.abs(X) < p['lim']) & (np.abs(Y) < p['lim'])
    Vs = vstar(d, tau); dpl = EPS + DV * tau
    fonts(fig_width=9)
    fig, (a, b) = plt.subplots(2, 1, figsize=(9, 10.5), gridspec_kw=dict(height_ratios=[2, 1.4]), sharex=True)
    a.plot(d, Vs, **VSTAR)
    for arm in ARMS:
        G, m, fwd, ch, _ = load('torus', arm, seed)
        V = chunked(fwd, torch.tensor(np.stack([X.ravel(), Y.ravel()], 1), dtype=torch.float32), torch.full((X.size,), tau)).numpy().reshape(X.shape)
        V = np.where(inside, V, np.nan); mean = np.nanmean(V, 1); std = np.nanstd(V, 1); mean[d <= EPS] = np.nan; std[d <= EPS] = np.nan
        a.plot(d, mean, color=STYLE[arm]['color'], ls=STYLE[arm]['ls'], lw=STYLE[arm].get('lw', p['lw']), zorder=STYLE[arm].get('zorder', 3)); a.fill_between(d, mean - std, mean + std, color=STYLE[arm]['color'], alpha=p['band_alpha'], lw=0, zorder=2)
        b.semilogy(d, std + 1e-9, color=STYLE[arm]['color'], ls=STYLE[arm]['ls'], lw=p['lw_err'], zorder=STYLE[arm].get('zorder', 3))
    for axx in (a, b): axx.axvline(dpl, color='0.5', ls=':', lw=1.5); axx.axvline(1.0, color='0.5', ls='-', lw=1)
    grid(b, 'both')
    if p['grid_left']: grid(a, 'major')
    if p['annotate']: a.text(dpl + 0.02, 0.9, 'cone: $d=\\varepsilon+\\Delta v\\,\\tau$\nplateau $V^*=\\tau$ beyond', fontsize=FS(11), color='0.3'); a.text(1.01, 0.25, '$d>1$: only the arcs\ninside the domain', fontsize=FS(10.5), color='0.3')
    a.set_ylabel(f'$\\langle V\\rangle_\\theta(d,\\tau={tau:g})$'); a.set_ylim(*p['ylim_value']); a.set_title('torus $\\mathbb{T}^{\\ \\,2}$', loc='left')
    b.set_ylim(*p['ylim_std']); b.set_ylabel('$\\sigma_\\theta[V](d)$'); b.set_xlabel('geodesic distance $d=\\|\\Delta\\|$'); b.set_xlim(0, np.sqrt(2))
    a.legend(handles=legend_handles(), loc='upper left')
    fig.tight_layout()
    if fname: fig.savefig(fname)
    plt.show()
angavg_fig(f'{FIGS}/fig_angavg_t2.pdf')
```

```python
def panel_census_rms(ax, p=P['census']):
    # value RMS of every run, grouped by topology (x), one marker per seed, colour = configuration
    rng = np.random.default_rng(0); S = _S()
    for gi, geom in enumerate(['circle', 'torus', 'torus4']):
        for ai, arm in enumerate(ARMS):
            rows = _rows(geom, arm); x = gi * 5 + ai + rng.uniform(-p['jitter'], p['jitter'], len(rows))
            ax.scatter(x, [o['value_rms'] for o in rows], color=STYLE[arm]['color'], marker=p['markers'][geom], s=p['marker_size'] * S ** 2,
                       alpha=p['marker_alpha'], edgecolor=p['edge'], lw=p['edge_lw'] * S, zorder=3)
    ax.set_yscale('log'); ax.set_ylim(*p['ylim_rms']); ax.set_xticks([gi * 5 + 1.5 for gi in range(3)]); ax.set_xticklabels(['$S^1$', '$\\mathbb{T}^{\\ \\,2}$', '$\\mathbb{T}^{\\ \\,4}$'])
    ax.tick_params(axis='x', which='minor', bottom=False, top=False)
    for gi in range(1, 3): ax.axvline(gi * 5 - 1, color='0.85', lw=plt.rcParams['grid.linewidth'])
    ax.set_ylabel('value RMS against $V^*$')
    if p['grid']: ax.grid(True, which='major', axis='y', color=plt.rcParams['grid.color'], lw=plt.rcParams['grid.linewidth'])
fonts(fig_width=P['census']['figsize'][0]); fig, ax = plt.subplots(figsize=(8, 5)); panel_census_rms(ax); plt.show()
```

```python
def panel_census_cert(ax, p=P['census']):
    # certificate max ratio against value RMS, every run
    S = _S()
    for geom in ['circle', 'torus', 'torus4']:
        for arm in ARMS:
            rows = _rows(geom, arm)
            ax.scatter([o['value_rms'] for o in rows], [o['cert_mc']['maxratio'] for o in rows], color=STYLE[arm]['color'], marker=p['markers'][geom],
                       s=p['marker_size'] * S ** 2, alpha=p['marker_alpha'], edgecolor=p['edge'], lw=p['edge_lw'] * S, zorder=3)
    ax.set_xscale('log'); ax.set_yscale('log'); ax.set_xlim(*p['xlim_rms']); ax.set_ylim(*p['ylim_ratio'])
    ax.axhline(p['threshold'], color='0.4', ls='--', lw=1.2 * S)
    ax.set_xlabel('value RMS against $V^*$'); ax.set_ylabel('$\\max\\,\\hat\\lambda_{\\max}/C^{*}$')
    if p['grid']: ax.grid(True, which='major', color=plt.rcParams['grid.color'], lw=plt.rcParams['grid.linewidth'])
fonts(fig_width=P['census']['figsize'][0]); fig, ax = plt.subplots(figsize=(6, 5)); panel_census_cert(ax); plt.show()
```

```python
# ---- assemble -> figs/<P['census']['fname']>: colour legend (configurations) and marker legend (topologies) above the panels
p = P['census']; fonts(fig_width=p['figsize'][0])
fig, (a, b) = plt.subplots(1, 2, figsize=p['figsize'], gridspec_kw=dict(width_ratios=p['width_ratios']))
panel_census_rms(a); panel_census_cert(b)
ms = 7.5 * _S()   # legend markers at the size of the plotted ones
Hc = [Line2D([], [], color=STYLE[k]['color'], marker='o', ls='', ms=ms, mec=p['edge'], mew=0.6 * _S(), label=STYLE[k]['label']) for k in ARMS]
Ht = [Line2D([], [], color='0.35', marker=p['markers'][g], ls='', ms=ms, mec=p['edge'], mew=0.6 * _S(), label=lab) for g, lab in [('circle', '$S^1$'), ('torus', '$\\mathbb{T}^{\\ \\,2}$'), ('torus4', '$\\mathbb{T}^{\\ \\,4}$')]]
H = [Hc[0], Ht[0], Hc[1], Ht[1], Hc[2], Ht[2], Hc[3]]   # column-major fill of a 2 x 4 legend: configurations on the first row, topologies on the second
fig.legend(handles=H, loc='upper center', ncol=p['legend_ncol'], bbox_to_anchor=(0.5, p['legend_y']), frameon=False, handlelength=1.0, handletextpad=0.4, columnspacing=1.4)
fig.tight_layout(rect=[0, 0, 1, p['top']]); fig.savefig(f"{FIGS}/{p['fname']}", bbox_inches='tight'); plt.show()
```
